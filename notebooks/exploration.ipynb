```python
import numpy as np

columns = {
    'title_fr' : 'Titre',
    'description_fr': 'Description',
    'longdescription_fr': 'Description Longue',
    'conditions_fr': 'Conditions',
    'keywords_fr': 'Mots clés',
    'accessibility_label_fr': 'Condition d\'accessibilité',
    'location_city': 'Ville',
    'location_department': 'Département',
}

formated_events = []

for index, event in events.iterrows():
    e = ''
    for alias, title in columns.items():
        try:
            if alias == 'longdescription_fr' and event[alias] is not None and event[alias] is not np.nan:
                cleaned = utils.clean_html(event['longdescription_fr'])
                e += f'{title}: {cleaned}, '
            elif alias in ['keywords_fr', 'accessibility_label_fr'] and event[alias] is not None and event[alias] is not np.nan:
                e += f'{title}: {' - '.join(event[alias])}, '
            else:
                e += f'{title}: {event[alias]}, '
        except Exception as ex:
            print(f'Error processing event {index} column {alias}: {ex}')

    formated_events.append(e)
```
